```python
features = anes_imputed.drop(columns=['democrat'], errors='ignore')
scaled = StandardScaler().fit_transform(features)
pca = PCA()
pca_fit = pca.fit(scaled)
```

```python
var_exp = pca.explained_variance_ratio_
cum_var_exp = np.cumsum(var_exp)
pc = np.arange(1, len(var_exp) + 1)
plt.figure(figsize=(10, 6))
sns.lineplot(x=pc, y=var_exp, marker='o')
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=pc, y=pca.explained_variance_)
```

```python
loadings = pd.DataFrame(pca.components_.T,  columns=[f'PC{i+1}' for i in range(pca.n_components_)], index=features.columns)

plt.figure(figsize=(12, 10))
sns.heatmap(loadings.iloc[:, :5], cmap="coolwarm", annot=True)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(data.drop(columns="democrat"))
```

```python
columns = data.drop(columns="democrat").columns
anes_scaled = pd.DataFrame(X_scaled, columns=columns)
anes_scaled["democrat"] = data["democrat"]
```

```python
color = anes_scaled['democrat'].map({0: 'red', 1: 'blue'}).astype(str).values
color_array = anes_scaled['democrat'].map({0: 'red', 1: 'blue'})
color_array = color_array.fillna('gray') 
rgb_colors = [mcolors.to_rgb(color) for color in color_array]

fig = plt.figure(figsize=(30, 20))
ax1 = fig.add_subplot(221, projection='3d')
ax1.scatter(anes_scaled['Trump'], anes_scaled['Obama'], anes_scaled['Sanders'],
            c=rgb_colors, s=10)
ax1.set_title("Politicians",  fontsize=30)
ax1.set_xlabel("Donald Trump")
ax1.set_ylabel("Barack Obama")
ax1.set_zlabel("Bernie Sanders")

ax2 = fig.add_subplot(222, projection='3d')
ax2.scatter(anes_scaled['NRA'], anes_scaled['NATO'], anes_scaled['UN'], c=rgb_colors, s=10)
ax2.set_title("Institutions", fontsize=30)
ax2.set_xlabel("NRA")
ax2.set_ylabel("NATO")
ax2.set_zlabel("UN")

ax3 = fig.add_subplot(223, projection='3d')
ax3.scatter(anes_scaled['Illegal'], anes_scaled['Immigrants'], anes_scaled['ICE'], c=rgb_colors, s=10)
ax3.set_title("Issues (Immigration)", fontsize=30)
ax3.set_xlabel("Illegal Immigrants")
ax3.set_ylabel("Immigrants")
ax3.set_zlabel("ICE")

ax4 = fig.add_subplot(224, projection='3d')
ax4.scatter(anes_scaled['Palestine'], anes_scaled['Saudi Arabia'], anes_scaled['Israel'], c=rgb_colors, s=10)
ax4.set_title("Countries (Middle East)", fontsize=30)
ax4.set_xlabel("Palestine")
ax4.set_ylabel("Saudi Arabia")
ax4.set_zlabel("Israel")
plt.tight_layout()
plt.show()
```

```python
results = []
for k in [3, 5, 7, 10, 15, 18, 19, 20]:
    lle = LLE(n_neighbors=k, n_components=2)
    dist = lle.fit_transform(X_scaled)
    orig = pairwise_distances(X_scaled)
    new = pairwise_distances(dist)
    
    rho = np.corrcoef(orig.flatten(), new.flatten())[0, 1]
    results.append((k, rho))
best_k, best_rho = max(results, key=lambda x: x[1])
print(f' k = {best_k}, corr = {best_rho}')
```

```python
lle= LocallyLinearEmbedding(n_neighbors=20, n_components=2)
lle_fit = lle.fit_transform(X_scaled)
lle_df = pd.DataFrame(lle_fit, columns=["Dim1", "Dim2"])
lle_df["Party"] = data["democrat"].map({0: "Non-Democrat", 1: "Democrat"})
```
